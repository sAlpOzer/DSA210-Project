First we need to determine which tier a Pokemon is in based on their usage in said tier. The rules that determine if a Pokemon is in a given tier can be found at the relevant smogon page for the tier. (For example OU: https://www.smogon.com/dex/sv/formats/ou/)

```python
def usage_based_filtering(tier_json):
  pkmn_in_tier = tier_json.copy()
  for pkmn, details in tier_json.items():
    usage = details.get('usage')['weighted']
    if usage < 0.0452: # Magic usage percent (determined by the smogon council) that determines if a Pokémon is in this tier
      del pkmn_in_tier[pkmn]
  return pkmn_in_tier
```

```python
def clear_duplicates(higher_tier, lower_tier):
  temp = higher_tier.copy()
  for pkmn, details in higher_tier.items():
    if pkmn in lower_tier:
      del temp[pkmn]
  return temp
```

Combine all the tiers into a one dataframe and add the labels (tier) column

```python
import pandas as pd

tier_dicts = {
    "Ubers": ubers_pkmn,
    "UUbers": uubers_pkmn,
    "OU": ou_pkmn,
    "UU": uu_pkmn,
    "RU": ru_pkmn,
    "NU": nu_pkmn,
    "PU": pu_pkmn,
    "ZU": zu_pkmn
}

combined_dataset = []

for tier, pkmn_data in tier_dicts.items():

    for name, details in pkmn_data.items():

        label_tier = "Ubers" if tier == "UUbers" else tier

        entry = {
            "Pokemon": name,
            "Tier": label_tier
        }

        # Extract abilities
        ability_list = list(details.get("abilities", {}).keys())
        for i in range(3):
            entry[f"Ability_{i+1}"] = ability_list[i] if i < len(ability_list) else None

        # Extract 10 most common moves
        move_list = list(details.get("moves", {}).keys())
        for i in range(10):
            entry[f"Move_{i+1}"] = move_list[i] if i < len(move_list) else None

        combined_dataset.append(entry)

df = pd.DataFrame(combined_dataset)

# Reorder columns to ensure tier (label) is on the rightmost column
cols = [c for c in df.columns if c != 'Tier'] + ['Tier']
df = df[cols]

print(df)
```

One-Hot encode roles as categorical columns based on the moves each Pokemon has, if they have a move listed in the role categories than they can perform that role

```python
# Role categories
hazard_moves = ["Stealth Rock", "Spikes", "Toxic Spikes", "Sticky Web", "Ceaseless Edge", "Stone Axe"]
pivot_moves = ["U-turn", "Volt Switch", "Flip Turn", "Parting Shot", "Chilly Reception", "Baton Pass"]
setup_moves = ["Swords Dance", "Dragon Dance", "Quiver Dance", "Calm Mind", "Nasty Plot", "Bulk Up", "Iron Defense", "Agility", "Shell Smash", "Victory Dance", "Tidy Up"]
recovery_moves = ["Recover", "Soft-Boiled", "Roost", "Slack Off", "Milk Drink", "Shore Up", "Synthesis", "Moonlight", "Morning Sun", "Wish", "Strength Sap", "Leech Seed"]
utility_moves = ["Rapid Spin", "Defog", "Mortal Spin", "Knock Off", "Taunt", "Encore", "Will-O-Wisp", "Thunder Wave", "Toxic", "Glare", "Trick", "Switcheroo", "Haze", "Roar", "Whirlwind"]
priority_moves = ["Sucker Punch", "Extreme Speed", "Jet Punch", "Aqua Jet", "Bullet Punch", "Ice Shard", "Mach Punch", "Shadow Sneak", "Thunderclap", "Vacuum Wave"]

categories = {
    "Hazard": hazard_moves,
    "Pivot": pivot_moves,
    "Setup": setup_moves,
    "Recovery": recovery_moves,
    "Utility": utility_moves,
    "Priority": priority_moves
}

for role, move_list in categories.items():
    categories[role] = [m.lower().replace(" ", "-") for m in move_list]
    print(f"{role}: {categories[role]}")
```

```python
move_cols = [f"Move_{i+1}" for i in range(10)]

for category_name, move_list in categories.items():
    # .isin() to check if any of the move columns contain a move from the category list
    # .any(axis=1) returns True if the move is found in any of the 10 columns
    # .astype(int) converts True/False to 1/0
    df[category_name] = df[move_cols].isin(move_list).any(axis=1).astype(int)

# Reorder to keep Tier at the end
cols = [c for c in df.columns if c != 'Tier'] + ['Tier']
df = df[cols]

print(df[['Pokemon', 'Hazard', 'Pivot', 'Setup', 'Recovery', 'Utility', 'Priority', 'Tier']].head())
```

```python
import pandas as pd

# list of all 18 types
all_types = [
    "normal", "fire", "water", "electric", "grass", "ice", "fighting", "poison",
    "ground", "flying", "psychic", "bug", "rock", "ghost", "dragon", "dark", "steel", "fairy"
]

def parse_type_matchups(type_json):
    """
    Creates a flat lookup: lookup[attacker][defender]
    """
    lookup = {t: {d: 1.0 for d in all_types} for t in all_types}

    for type_name, data in type_json.items():
        rel = data['damage_relations']

        for item in rel['double_damage_to']:
            lookup[type_name][item['name']] = 2.0
        for item in rel['half_damage_to']:
            lookup[type_name][item['name']] = 0.5
        for item in rel['no_damage_to']:
            lookup[type_name][item['name']] = 0.0

    return lookup

type_lookup = parse_type_matchups(pokeapi_types_json)
```

Assign a score to the defensive and offensive properties of each Pokemon's type combinations instead of providing the raw types as features

```python
def calculate_type_metrics(pkmn, pkmn_type_map):

    if "arceus" in pkmn and "-" in pkmn:
        t1 = pkmn.split("-")[1].lower()
        t2 = None
    else:
      types = pkmn_type_bs_map.get(pkmn.lower())[0]
      # types is expected to be a list: ['dark', 'steel']
      t1 = types[0].lower()
      t2 = types[1].lower() if len(types) > 1 and types[1] else None

    def_score = 0
    off_score = 0

    # --- Defensive Score ---
    for attacker in all_types:

        mult = type_lookup[attacker][t1]

        if t2:
            mult *= type_lookup[attacker][t2]

        # Numbers can be changed depending on importance
        # Personally, I believe immunity's are significantly more valuable than x4 resists
        # Similarly 4x weaknesses are too much of a liability which in my mind justifies the harsh score penalization
        if mult == 0:    def_score += 6  # Immunity
        elif mult == 0.25: def_score += 4  # 4x Resist
        elif mult == 0.5:  def_score += 3  # 2x Resist
        elif mult == 1:    def_score += 0.5 # Neutral Hit
        elif mult == 2:    def_score -= 3  # x2 Weakness
        elif mult == 4:    def_score -= 8  # 4x Weakness

    # --- Offensive Score ---
    for target in all_types:
        # How effective Offensively is a Pokemon Same Type Attack Bonus(STAB) types are
        m1 = type_lookup[t1][target]
        m2 = type_lookup[t2][target] if t2 else 0
        best_mult = max(m1, m2)

        if best_mult == 2:   off_score += 2 # x2 Dmg
        elif best_mult == 1: off_score += 0.5 # x1 Dmg
        elif best_mult == 0.5: off_score -= 2 # x0.5 Dmg
        elif best_mult == 0:  off_score -= 4 # x0 Dmg

    return def_score, off_score
```

```python
def create_name_to_type_bs_map(pokemon_api_json):
    name_map = {}
    for key, data in pokemon_api_json.items():

        types = [t['type']['name'] for t in data.get('types')]
        baseStats = [s['base_stat'] for s in data.get('stats')]

        name_map[key] = types, baseStats
    return name_map

pkmn_type_bs_map = create_name_to_type_bs_map(pokeapi_pkmn_json)
```

Speed is often considered a threshold stat, so instead of giving the model raw speeds, provide speed as tiers

```python
def get_speed_tier(row):
  speed = row['Spe']

  if speed <= 60:
    return 4
  elif speed <= 100:
    return 3
  elif speed <= 130:
    return 2
  else:
    return 1
```

```python
def get_phys_bulk_score(row):
  HP = row['HP']
  defense = row['Def']

  return HP * defense
```

```python
def get_spe_bulk_score(row):
  HP = row['HP']
  spe_def = row['SpD']

  return HP * spe_def
```

```python
def get_atk_to_spa_ratio(row):
  atk = row['Atk']
  spa = row['SpA']

  return atk / spa
```

```python
def get_wasted_attacking_stat_ratio(row):
  stat_list = []

  stat_list.append(row['HP'])
  stat_list.append(row['Atk'])
  stat_list.append(row['Def'])
  stat_list.append(row['SpA'])
  stat_list.append(row['SpD'])
  stat_list.append(row['Spe'])

  atk = row['Atk']
  spa = row['SpA']

  bst = sum(stat_list)

  if (atk > spa):
    return spa/bst
  else:
    return atk/bst
```

```python
df['SpeedTier'] = df.apply(get_speed_tier, axis=1)
df['PhysBulk'] = df.apply(get_phys_bulk_score, axis=1)
df['SpeBulk'] = df.apply(get_spe_bulk_score, axis=1)
df['AtkToSpaRatio'] = df.apply(get_atk_to_spa_ratio, axis=1)
df['WastedStatRatio'] = df.apply(get_wasted_attacking_stat_ratio, axis=1)

# Ensure Tier remains the last column
cols = [c for c in df.columns if c != 'Tier'] + ['Tier']
df = df[cols]

print(df[['Pokemon', 'PhysBulk', 'SpeBulk', 'AtkToSpaRatio', 'WastedStatRatio' ,'SpeedTier', 'Tier']])
```
