```python
import requests
# Get the information of a select Pokemon
def get_pokemon_details(pokemon_name):
    url = f"https://pokeapi.co/api/v2/pokemon/{pokemon_name.lower()}"

    try:
        response = requests.get(url)
        if response.status_code == 200:
            data = response.json()

            # 1. Name
            print(f"\n{'='*30}")
            print(f" {data['name'].upper()} ")
            print(f"{'='*30}")

            # 2. Types
            types = [t['type']['name'] for t in data['types']]
            print(f"Type(s): {', '.join(types)}")

            # 3. Abilities
            print("\nAbilities:")
            for entry in data['abilities']:
                ability_name = entry['ability']['name'].replace('-', ' ').title()
                hidden_tag = "(Hidden)" if entry['is_hidden'] else ""
                print(f" • {ability_name} {hidden_tag}")

            # 4. Stats
            print("\nBase Stats:")
            for s in data['stats']:
                stat_name = s['stat']['name'].replace('-', ' ').upper()
                stat_value = s['base_stat']
                print(f" {stat_name:<15}: {stat_value:<3}")

            print(f"{'='*30}\n")

        elif response.status_code == 404:
            print("Error: Pokémon not found.")

    except Exception as e:
        print(f"Connection error: {e}")

name_input = input("Enter Pokémon name: ")
get_pokemon_details(name_input)
```

```python
import requests
# Get the names of all generation 5 Pokemon in Pokedex order
def get_gen_5_list():

    url = "https://pokeapi.co/api/v2/pokemon?limit=156&offset=493"

    try:
        response = requests.get(url)
        if response.status_code == 200:
            data = response.json()
            pokemon_list = data['results']

            print(f"{'ID':<5} | {'Name':<15}")
            print("-" * 22)

            for index, pokemon in enumerate(pokemon_list, start=1):
                name = pokemon['name'].capitalize()
                print(f"#{index:<4} | {name:<15}")

            print(f"\nTotal fetched: {len(pokemon_list)}")
        else:
            print(f"Failed to retrieve data. Status code: {response.status_code}")

    except Exception as e:
        print(f"An error occurred: {e}")

get_gen_5_list()
```
